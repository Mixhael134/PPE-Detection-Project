```python
from ultralytics import YOLO
import matplotlib.pyplot as plt
import cv2
```

```python
model = YOLO('/content/runs/detect/ppe_detection_model-2/weights/best.pt')
```

```python
!ls -F /content/CompVis_Data/images/

# Replace 'your_actual_image_name.jpg' with a valid image file name from the output above
results = model.predict(
    source='/content/CompVis_Data/images/pexels-photo-10039979.jpeg',
    save=True,
    conf=0.25
)
```

```python
import os

if results and len(results) > 0:
    # Get the annotated image as a NumPy array directly from the results object
    annotated_image = results[0].plot()

    if annotated_image is None:
        print("Error: The plot() method did not return an image. This might indicate an issue with the prediction results.")
    else:
        plt.figure(figsize=(12,12))
        # OpenCV images are BGR by default, matplotlib expects RGB
        plt.imshow(cv2.cvtColor(annotated_image, cv2.COLOR_BGR2RGB))
        plt.axis('off')
        plt.title('PPE Detection Demo Result')
else:
    print("Error: No prediction results found. Please run the prediction cell first.")
```
