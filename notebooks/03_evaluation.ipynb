```python
from ultralytics import YOLO
import matplotlib.pyplot as plt
import cv2
```

```python
model = YOLO('/content/runs/detect/ppe_detection_model-2/weights/best.pt')
```

```python
metrics = model.val()

print(metrics)
```

```python
image = cv2.imread('/content/runs/detect/ppe_detection_model-2/confusion_matrix.png')

plt.figure(figsize=(10,10))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Confusion Matrix')
```

```python
image = cv2.imread('/content/runs/detect/ppe_detection_model-2/results.png')

plt.figure(figsize=(14,8))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Training Results')
```

```python
image = cv2.imread('/content/runs/detect/ppe_detection_model-2/val_batch0_pred.jpg')

plt.figure(figsize=(12,12))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Validation Prediction Example')
```
